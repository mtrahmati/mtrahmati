## 3. Use it from Python
Pass any Python function and the limits.

```python
import math
from integrate import integrate

print(integrate(lambda x: x**2, 0, 1))            # 1/3
print(integrate(math.sin, 0, math.pi))              # 2
print(integrate(lambda x: math.exp(-x**2), -10, 10), math.sqrt(math.pi))
```

### Compare the three methods

```python
f, a, b, exact = math.sin, 0, math.pi, 2.0
for method in ["trapezoid", "simpson", "adaptive"]:
    r = integrate(f, a, b, method)
    print(f"{method:10s} {r:.15f}   error = {abs(r - exact):.1e}")
```
